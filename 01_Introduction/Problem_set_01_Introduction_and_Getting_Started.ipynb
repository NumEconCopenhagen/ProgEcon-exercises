## 2. Calculate monthly transport costs

A student travels to campus on 18 days during a month and pays for two trips on each travel day. A single trip costs 24 DKK. In the following month, the fare rises by 12.5 percent and the student travels on 16 days.

1. Calculate the original monthly cost.
2. Calculate the new fare without overwriting the original fare.
3. Calculate the new monthly cost using 16 travel days.
4. Calculate the change in monthly cost.
5. Calculate what the monthly cost would have been if the fare had increased but the student had still travelled on 18 days. Store it as `price_only_cost`.
6. Print the results with clear labels.

In a short Markdown note, explain why `price_only_cost` differs from the new monthly cost.

````python
single_trip_fare = 24.0
travel_days = 18
trips_per_day = 2
fare_increase_rate = 0.125
new_travel_days = 16
````

**Explanation:**

<details>
<summary><strong>Answer</strong></summary>

```python
single_trip_fare = 24.0
travel_days = 18
trips_per_day = 2
fare_increase_rate = 0.125
new_travel_days = 16

original_monthly_cost = travel_days * trips_per_day * single_trip_fare
new_single_trip_fare = single_trip_fare * (1 + fare_increase_rate)
new_monthly_cost = new_travel_days * trips_per_day * new_single_trip_fare
change_in_monthly_cost = new_monthly_cost - original_monthly_cost
price_only_cost = travel_days * trips_per_day * new_single_trip_fare

print("Original monthly cost:", original_monthly_cost, "DKK")
print("New fare per trip:", new_single_trip_fare, "DKK")
print("New monthly cost:", new_monthly_cost, "DKK")
print("Change in monthly cost:", change_in_monthly_cost, "DKK")
print("Cost with only the fare change:", price_only_cost, "DKK")
```

`price_only_cost` keeps 18 travel days, while the new monthly cost uses 16. The higher fare therefore applies to fewer trips.

Notice that the change in monthly cost comes out as exactly `0.0`. Your code is fine. The fare rise adds 108 DKK at 18 travel days (864 to 972), and cutting the month to 16 travel days takes the same 108 DKK away again (972 back to 864). Separating a change into the part caused by the price and the part caused by the quantity is something you will do again and again in economics; here the two happen to cancel exactly.
